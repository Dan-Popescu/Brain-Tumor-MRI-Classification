```python
# VERSION PYSPARK (nécessite Java 17 ou inférieur)
from pyspark.sql import SparkSession
from IPython.display import display


spark = SparkSession.builder.getOrCreate()

versions_df = spark.read.parquet("../data/processed/transform_versions.parquet")
display(versions_df.orderBy("generated_at_utc", ascending=False).toPandas())
```

```python
# VERSION PYSPARK
split_versions_df = spark.read.parquet("../data/processed/split_versions.parquet")
display(split_versions_df.orderBy("generated_at_utc", ascending=False).toPandas().T)
```
